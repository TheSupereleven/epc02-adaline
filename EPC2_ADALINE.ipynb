```python

import random
import pandas as pd
import matplotlib.pyplot as plt

eta = 0.0025
epsilon = 1e-6
max_epocas = 10000

def calcular_eqm(x, d, pesos):
    soma = 0

    for i in range(len(x)):
        entradas = x.iloc[i].tolist()
        entradas_bias = [-1.0] + entradas
        desejada = float(d.iloc[i])

        u = sum(p * x for p, x in zip(pesos, entradas_bias))
        erro = desejada - u
        soma += erro ** 2

    return soma / len(x)


def treinar_adaline(x, d, pesos_iniciais):
    pesos = pesos_iniciais.copy()
    historico_eqm = []

    for epoca in range(max_epocas):
        for i in range(len(x)):
            entradas = x.iloc[i].tolist()
            entradas_bias = [-1.0] + entradas
            desejada = float(d.iloc[i])

            u = sum(p * x for p, x in zip(pesos, entradas_bias))
            erro = desejada - u

            pesos = [
                p + eta * erro * x
                for p, x in zip(pesos, entradas_bias)
            ]

        eqm = calcular_eqm(x, d, pesos)
        historico_eqm.append(eqm)

        if len(historico_eqm) > 1:
            diferenca = abs(historico_eqm[-2] - historico_eqm[-1])

            if diferenca < epsilon:
                break

    return pesos, len(historico_eqm), historico_eqm


# Executa os cinco treinamentos
resultados = []

for numero in range(5):
    pesos_iniciais = [random.uniform(0, 1) for _ in range(5)]

    pesos_finais, epocas, historico_eqm = treinar_adaline(
        x, d, pesos_iniciais
    )

    resultados.append({
        "treinamento": numero + 1,
        "pesos_iniciais": pesos_iniciais,
        "pesos_finais": pesos_finais,
        "epocas": epocas,
        "eqm_final": historico_eqm[-1],
        "historico_eqm": historico_eqm
    })

# Tabela de resultados
linhas = []

for r in resultados:
    linha = {
        "Treinamento": r["treinamento"],
        "Épocas": r["epocas"],
        "EQM final": r["eqm_final"]
    }

    for i in range(5):
        linha[f"w{i}_inicial"] = r["pesos_iniciais"][i]
        linha[f"w{i}_final"] = r["pesos_finais"][i]

    linhas.append(linha)

tabela_resultados = pd.DataFrame(linhas)
display(tabela_resultados.round(8))
```
